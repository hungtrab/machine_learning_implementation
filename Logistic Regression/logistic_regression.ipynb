```python
import numpy as np
import pandas as pd
```

```python
data_path = 'data/breast-cancer.csv'
```

```python
data = pd.read_csv(data_path)
data['diagnosis'] = data['diagnosis'].map({'M': 1, 'B': 0})
features = data.drop('diagnosis', axis=1)
labels = data['diagnosis']
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2, random_state=42)
```

```python
# With data preprocessing
# Scaling numerical features
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
numerical_features = features.select_dtypes(include=[np.number]).columns
features[numerical_features] = scaler.fit_transform(features[numerical_features])
labels = data['diagnosis']
from sklearn.model_selection import train_test_split
X_scale_train, X_scale_test, y_scale_train, y_scale_test = train_test_split(features, labels, test_size=0.2, random_state=42)
```

```python
def binary_cross_entropy(y_true, y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    bce = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    return bce
```

```python
class LogisticRegression:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.theta = None

    def _sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        self.theta = np.zeros(X_b.shape[1])

        for _ in range(self.n_iterations):
            linear_model = X_b.dot(self.theta)
            y_pred = self._sigmoid(linear_model)
            gradients = X_b.T.dot(y_pred - y) / y.size
            self.theta -= self.learning_rate * gradients

    def predict(self, X):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        linear_model = X_b.dot(self.theta)
        y_pred = self._sigmoid(linear_model)
        return (y_pred >= 0.5).astype(int)
```

```python
def accuracy(y_true, y_pred):
    return np.mean(y_true == y_pred)
```

```python
model = LogisticRegression(learning_rate=0.01, n_iterations=1000)
model.fit(X_scale_train, y_scale_train)
y_pred = model.predict(X_scale_test)
acc = accuracy(y_scale_test, y_pred)
print(f"Logistic Regression Accuracy: {acc}")
```

```python
model_1 = LogisticRegression(learning_rate=0.01, n_iterations=1000)
model_1.fit(X_train, y_train)
y_pred_1 = model_1.predict(X_test)
acc_1 = accuracy(y_test, y_pred_1)
print(f"Logistic Regression Accuracy without scaling: {acc_1}")
```
